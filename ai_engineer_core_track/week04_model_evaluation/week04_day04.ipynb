````python
# Now, since we have to build this in gradio, our functions must be designed according to that also. First let's finalise the app design.
# 1. row one to have two parts, one for showing the existing code, and second to show the code that gets ported. Basically, the python code on the left and converted c++ code on the right.
# 2. Then we will have another row, where, we will have the buttons and drop downs: 1. to run python code, 2. to port python code to c++ (for now, we might add more language options later), 3. To run c++ code and finally and finally we will have a drop down to select from the given LLM options.
# 3. Again, we will have one row, with two rows, where, we will have 1. the output of the python code and 2. the output of the c++ code. 

# We can also store the results in sqlite, but this is something we can work on later, since this would need us to create a new page with the stored data. For now, we work on the 3 rows. 

# Now, for these 3 rows to work as is, we need the following:
# 1. A function, that returns the python code as the output. > For now, this shall remain static and loaded by default.
# 2. A function, that runs this python code and returns the output. > Run Python button runs this function and its output is loaded to python output section.
# 3. A function, that ports the given python code and returns the C++ code. > port python runs this > and its output is displayed in the show c++ code. In the background, it will also save the code to the file. For this, the input will be the model name.
# 4. A function to compile and run the c++ code. The input of this will be the compiled file. Output will be displayed in the c++ output section. 
````

````python
# we start by putting in the python code

pi = """
import time

def calculate(iterations, param1, param2):
    result = 1.0
    for i in range(1, iterations+1):
        j = i * param1 - param2
        result -= (1/j)
        j = i * param1 + param2
        result += (1/j)
    return result

start_time = time.time()
result = calculate(200_000_000, 4, 1) * 4
end_time = time.time()

print(f"Result: {result:.12f}")
print(f"Execution Time: {(end_time - start_time):.6f} seconds")
"""
````

````python
# Now, we need to show the code without the """ """, we need and return that from the function


def show_python(python):
    clean_code = python.removeprefix(('"""').removesuffix('"""'))
    return clean_code
````

````python
# next we move to run python, since, the code must be run also and its output needs to be displayed

def run_python(python):
    python_output = exec(python)
    return python_output
````

````python
# next up is system_info function

from mac_sysinfo import get_unified_m2_context
````

````python
def system_info():
    system_info = get_unified_m2_context(as_json=True)
    return system_info
````

````python
# now lets add the compile message
compile_message = ["clang++", "-std=c++17", "-Ofast", "-mcpu=native", "-flto=thin", "-fvisibility=hidden", "-DNDEBUG", "main.cpp", "-o", "main"]
````

````python
# now, we move towards creating the messages section
def return_messages(python=pi, compile_info=compile_message):
    sys_info = system_info()
    system_message = f"""
        Your task is to convert Python code into high performance C++ code. Respond only with C++ code. Any extra explanation or details that you want to share, must be in the shape of C++ comments.
        The C++ response needs to produce an identical output in the fastest possible time. The code must be optimized to run on the specified system details mentioned below:
        {sys_info}
        """
    user_message = f"""
        Port this Python code to C++ with the fastest possible implementation that produces identical output in the least time.
        The system information is:
        {sys_info}
        Make sure that your code uses maximum use of the hardware and executes the code in the fastest possible time. You should also make maximum use of the cores in the hardware, to create code that can run simultaneously. Note that code being super fast is our main objective. You can change the algorithm as well if needed, we are not looking for translation of the code, but a code that does exactly what the below code achieved. 
        Your response will be written to a file called main.cpp and then compiled and executed; the compilation command is:
        {compile_info}
        Respond only with C++ code.
        Python code to port:
    
        ```python
        {python}
        ```
        """

    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ]

    return messages
````

````python
# next we add the extract c++ code, which we will be using to extract the c++ code from the LLM response. This we are using from the previous file and same goes for all the following c++ processing functions.

import re
from pathlib import Path
````

````python
def extract_cpp(llm_output):
    """Pull raw C++ out of whatever the model returned."""
    # Grab the contents of the first ```...``` block, if there is one.
    match = re.search(r"```(?:cpp|c\+\+)?\s*(.*?)```", llm_output, flags=re.DOTALL | re.IGNORECASE)
    code = match.group(1) if match else llm_output   # fallback: no fence → use it all
    return code.strip() + "\n"                        # trailing newline = good source hygiene
````

````python
# now we add the code for writing the file
def write_cpp_file(llm_output, filename="main.cpp"):
    """Clean the model output and write it to a .cpp file. Returns the path."""
    code = extract_cpp(llm_output)
    Path(filename).write_text(code, encoding="utf-8")
    return filename
````

````python
# so, these shall suffice our port function, where the task is to give the code as output and save it to a file. this code takes model as the input. Also, this is for now configured only for open router, we will configure this for a small model that we shall run on our system later.

def port_code(run_model):
    messages = return_messages()
    client = OpenAI(api_key=open_router_key, base_url=open_router_url)
    response = client.chat.completions.create(model=run_model, messages=messages, max_tokens=32000)
    llm_content = response.choices[0].message.content
    if llm_content:
        c_code = extract_cpp(llm_content)
        write_cpp_file(c_code, filename="llm_produced_code/day04/main.cpp")
        return c_code
    else:
        return "Empty response received from LLM. Please check LLM settings."
````

````python
# now, port is working find, in terms of producing the c++ code and saving it to a file. next we move to compile and run c++ code function.
import subprocess
# # next, lets work on the model list
# # DeepSeek V4 Pro, Kimi K3, DeepSeek V4.1 Flash, GLM-5.3, GLM-5.3 Flash, GPT OSS 20, GPT OSS 120

# model_list = ['DeepSeek-V4-Pro', 'Kimi-K3', 'DeepSeek-V4.1-Flash', 'GLM-5.3', 'GLM-5.3-Flash', 'GPT-OSS-20', 'GPT-OSS-120']

````

````python
def compile_and_run(source_file="llm_produced_code/day04/main.cpp", executable='main', timeout=120):
    """Compile an existing .cpp file and run it.

    Assumes the file is already written (e.g. by write_cpp_file).
    Returns {"stage", "success", "output", "error"} so the caller can react
    to *where* it failed (compile vs run).
    """
    # Guard: make sure the source file is actually there before we try.
    if not Path(source_file).exists():
        return {"stage": "setup", "success": False, "output": "",
                "error": f"Source file not found: {source_file}"}

    # 1. Compile. stderr is where compile errors go — capture it so we can see
    #    (or later feed back to the model) exactly what failed.
    # compile_cmd = ["clang++", "-std=c++17", "-O3", "-DNDEBUG", source_file, "-o", executable]
    compile_cmd = ["clang++", "-std=c++17", "-Ofast", "-mcpu=native", "-flto=thin", "-fvisibility=hidden", "-DNDEBUG", "llm_produced_code/day04/main.cpp", "-o", "main"]
    compiled = subprocess.run(compile_cmd, capture_output=True, text=True)
    if compiled.returncode != 0:
        return {"stage": "compile", "success": False, "output": "", "error": compiled.stderr}

    # 2. Run the executable. "./" tells the shell it's a local file.
    #    timeout guards against a bad port that hangs in an infinite loop.
    try:
        ran = subprocess.run([f"./{executable}"], capture_output=True, text=True, timeout=timeout)
    except subprocess.TimeoutExpired:
        return {"stage": "run", "success": False, "output": "", "error": f"Timed out after {timeout}s"}

    if ran.returncode != 0:
        return {"stage": "run", "success": False, "output": ran.stdout, "error": ran.stderr}

    return {"stage": "done", "success": True, "output": ran.stdout, "error": ""}
````

````python
# lets run the code and test it 

compile_and_run()
````

````python
# now since exec() returns None, we need to add code that returns, since this code is currently out of scope of my python understanding, getting this from ed's code
import sys
import io
def run_python_code_again(code):
    globals_dict = {"__builtins__": __builtins__}

    buffer = io.StringIO()
    old_stdout = sys.stdout
    sys.stdout = buffer

    try:
        exec(code, globals_dict)
        output = buffer.getvalue()
    except Exception as e:
        output = f"Error: {e}"
    finally:
        sys.stdout = old_stdout

    return output
````

````python
# great, the file runs and the best part is that it runs much faster than the code that was ported using frontier models such as astra, fable etc. However, there is a very small variation in the result. 

# now, moving ahead, we start working on the gradio part
llm_list = ['deepseek/deepseek-v4-pro-0813', 'moonshotai/kimi-k3', 'deepseek/deepseek-v4.1-flash', 'z-ai/glm-5.3', 'z-ai/glm-5.3-flash', 'openai/gpt-oss-20b', 'openai/gpt-oss-120b', 'z-ai/glm-5.2:free', 'deepseek/deepseek-v4-flash-0731:free', 'qwen/qwen3.8-27b:free']

with gr.Blocks() as language_port:
    gr.Markdown("Welcome to Code Port Machine")
    with gr.Row():
        python_code = gr.Textbox(value=pi, label='Python Code')
        c_plus = gr.Textbox(label='Converted C++ Code', lines=19)
    with gr.Row(equal_height=True):
        run_python_code = gr.Button(value='Run Python')
        select_llm = gr.Dropdown(llm_list, value=llm_list[0], show_label=False)
        port_to_c_plus = gr.Button(value='Convert to C++')
        run_c_plus_code = gr.Button(value='Run C++')
    with gr.Row():
        python_output = gr.Textbox(label='Python Code Output', lines=5)
        c_plus_output = gr.Textbox(label='C++ Code Output', lines=5)

    run_python_code.click(fn=run_python_code_again, inputs=[python_code], outputs=[python_output])
    port_to_c_plus.click(fn=port_code, inputs=[select_llm], outputs=[c_plus])
    run_c_plus_code.click(fn=compile_and_run, outputs=[c_plus_output])



language_port.launch()
````
