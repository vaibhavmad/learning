#### How to evaluate the performance of a Gen AI solution?

1. Model centric or technical metrics: loss, perplexity, accuracy, auto-roc, etc. are covered under this category and primarily used for training time model evaluation. These shall be covered later in the course when we do fine tuning during week 6 and week 7.

2. Business centric or outcome centric: KPIs, business objectives, customer satisfaction, etc. Basically how much business performance has been improved or what all objectives are being met.
