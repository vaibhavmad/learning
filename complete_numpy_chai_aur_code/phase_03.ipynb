```python
import numpy as np
```

```python
# we start by using a dataset to practice some more commands.

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array(
    [
        [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
        [2, 120000, 140000, 160000, 190000],  # Beijing Bites
        [3, 200000, 230000, 260000, 300000],  # Pizza Hub
        [4, 180000, 210000, 240000, 270000],  # Burger Point
        [5, 160000, 185000, 205000, 230000],  # Chat Point
    ]
)
```

```python
# now on this small dataset, we shall run operations basis certain needed outputs.

# 1. first, we want to print out the sales data of first 3 restaurants. For this, we need to print entire columns of rows at index 0, 1 and 2. So, here, we need to run slicing. For slicing, we remember if only row is to be printed, then we simply pass in the row index and since, we do not need any specific columns, we keep that empty

print(f"Complete sales data of first 3 restaurants:\n{sales_data[0:3,]}")
```

```python
# next we need to find total sales per year. Now, to do this, for every year, we need to sum the columns and present that as an array. Now, we know, for sum, we can use sum function. However, we need to take each column sum, so, the data must run from top to bottom, therefore, we need to use axis 0 to do this and also, we do not need the first column. Let us try running it below:

print(f"Total sale for each year:\n{np.sum(sales_data[:, 1:], 0)}")
```

```python
# next up, we need total sales for all restaurants, in this case, we use axis as 1, since we need to sum from left to right, and we sum up each column, except the first one or we can also say, we sum up the entire row and all columns

print(f"Total sale per restaurant: {np.sum(sales_data[:, 1:], 1)}")
```

```python
# next we need to find, minimum sales per restaurant, for this we use min, and since it is per row, we use index as 1, since data is flowing from left to right and except the first column, we need to check for all in that row

print(f"Minimum sale per restaurant is:\n{np.min(sales_data[:, 1:], 1)}")
```

```python
# similarly we can find maximum sale

print(f"Maximum sale per restaurant is:\n{np.max(sales_data[:, 1:], 1)}")
```

```python
# next up, we can find the average of each year, we use mean for this, and axis shall be 0. we check for each column, but not the first one.

print(f"Average sale per year:\n{np.mean(sales_data[:, 1:], 0)}")
```

```python
# and if we need to find average per restaurant, we do the following

print(f"Average sale per restaurant is:\n{np.mean(sales_data[:, 1:], 1)}")
```
