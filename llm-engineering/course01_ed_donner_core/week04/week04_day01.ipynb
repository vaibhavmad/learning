### 1. Choosing the Right LLM: Model Selection Strategy and Basics

The first thing Ed starts this week is to highlight that there is no best LLM. We need to identify actually what is the best LLM for a given task at hand. Now,  to decide the best model for the task at hand, depends upon task at hand. 

#### To compare LLMs, we first need to compare the basics such as:
1. Open Source or Closed?
2. Chat / Reasoning / Hybrid > Important, since reasoning models are slow and also are not very good at creative tasks. Also, there are hybrid models, which can select whether reasoning is needed or not depending upon the task at hand. And sometimes, we may not want a hybrid model, since we only want chat in the product.
3. Release Date and Knowledge Cut Off > This is an important aspect since the model will be aware of information that came before its knowledge cut off date. A model which was trained till july 24, it will not know anything post this date. For this, we will have to either provide it with tools or information via prompts, if the agent must be aware of this information.
4. Parameters > More parameters means better prediction, but we should not over focus on this since with smaller models people have found good results also.
5. Training Tokens > This translates to amount of training data. More training data means more knowledge, so the more the better.
6. Context Window > Total capacity a model has to look back on the conversation, and depends on the task at hand, if history is needed for the task, how much of it is needed, etc.

All this information can be found in the model cards that each provider publishes for their models.

7. Inference cost
8. Training cost > If fine tuning is needed to get the results.
9. Build cost > Do we also need to add more infra such as RAG etc?
10. Time to Market > How long will it take us to build the product.
11. Rate limits > How many times we can call a model in a given unit of time. 
12. Speed > Models have different characteristics when it comes to speed. Also, this is kind of inversely proportional to accuracy in my opinion.
13. Latency > How much time the model takes to generate the first token.
14. License > If we are working with open source, we need to also figure out if we can actually use the models for commercial goals. Some models can not be used commercially or some models have revenue limits for a commercial product.

This is something that we have to work on by ourselves.

### 2. The Chinchilla Scaling Law: Parameters, Training Data and Why It Matters

This was proposed by google. The idea of this law > They were trying to figure out how do we scale from model A to model B to by increasing its parameters while also increasing its performance. Example > scaling from a model with 8B parameters to a 16B parameters while doubling it's performance also.

What they discovered > You need to scale the training data in the same proportions, which means you also need to double the training data when we consider the above example.

In other words:

<b>Number of parameters ~ Number of training tokens</b>

This also means, that if you train a model with more training data and see very less improvement in model intelligence, this becomes a sign for increasing the parameters. 

Also, this is becoming less important now since there are ways to train models using better techniques, better architecture, better inference time techniques. But still important to understand.

### 3. Understanding AI Model Benchmarks: GPQA, MMLU-Pro, and HLE

Once we have gone through the requirements above, we can go through the benchmarks. Also, benchmarks are not the final word, but give us an idea of the capabilities of a model. Below are some of the most used benchmarks.

1. GPQA: Google proof question answers. Contains a list of 448 expert questions of PHD level which can not answered by searching on google. 65% score has been achieved by PHD level humans and 34% is scored by non PHD individuals.
2. MMLU-PRO: Judges models on language understanding. It stands for Massive Multitask Language Understanding 
3. AIME: This is for maths, a competitive maths puzzle test taken by top math schools for high schoolers and is invite only.
4. LiveCode Bench: As the name suggests, it evaluates the model's ability to work with code.
5. MuSR: This is used to judge a model's reasoning capabilities. It stands for Multi Step Soft Reasoning. Uses questions which require logical deduction like a murder mystery and answering Who has means, motive and opportunity.
6. HLE: Humanity's Last Exam is considered one of the hardest test created by human beings and is used to test Super human intelligence. When this came out in late 2024, models were scoring ~ 3% but now have reached ~60% with claude Fable 5.1


### 4. Limitations of benchmarks: Data contamination and over-fitting

There are also some very serious limitations of benchmarks. Therefore, we must treat benchmarks as an indication.

1. Training data contamination: Information on what the models are being tested gets leaked and models are trained on the questions and answers. Therefore as a result, the models perform better. To do this, benchmark providers keep rotating the information. Apple ran an experiment, where they changed the figures while keeping the same context in the questions. For example: changing the numbers in a math problem, while keeping the problem same. And not surprisingly a lot of agents scored way less.

2. Not consistently applied: Different models are not evaluated in the same manner.

3. Too narrow in scope: If a model is good at physics, chemistry and maths, but can be not that great at other subjects, therefore a benchmark checking for physics, chemistry or maths will do great at this. However, it does not mean that it will be great at other subjects also. One such agent was asked to create an image of a chess board one step away from a check-mate and it was not able to do so.

4. Hard to measure nuanced reasoning: A lot of benchmarks have multiple choice answers, therefore giving options right away.

5. Saturation: Models are becoming better and better and some benchmarks where models score 99% or above, wont be able to further test newer models with improved capabilities.

6. Over-fitting: Sometimes providers come out with 3 to 4 versions of the next gen and pick up the models which are great at a specific benchmark. This is also not a great way to proceed for the providers since their models will become that benchmark specialist in the longer run while loosing out on other aspects. 


<strong>A newer issue that is yet to be proven is that LLMs already have a understanding that they are being evaluated, thus models align(behave better or follow instructions better) better during benchmarks while not doing so when in production.</strong>
