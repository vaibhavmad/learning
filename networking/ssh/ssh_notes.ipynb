# SSH

SSH stands for secure shell. Basically, it is a way for two systems to interact with each other over un-secured connection. It uses special cryptography or we can say encoding of information that can not be understood by an intercepting party.

### What is it used for?
Before SSH was introduced, or being used we had data transfer mechanisms which used to send the password and username details directly in plain text. SSH, however, uses special encoding, which can not be read by an intercepting party till the time they do not have the decoding ability of the encoded information.

Below are some of it's uses:
- **Remote Command-Line Execution**: Logging into a remote server to run commands as if you were sitting directly in front of that machine.
- **Secure File Transfer**: Moving files safely between computers using associated protocols like SFTP (SSH File Transfer Protocol) or SCP (Secure Copy).
- **Port Forwarding (Tunneling)**: Routing local network traffic through the encrypted SSH connection to safely access databases, bypass firewalls, or secure otherwise unencrypted applications.

### How SSH keeps data secured?
The protocol uses a combination of three cryptographic techniques to protect the data being sent from eavesdropping and hijacking:
- **Asymmetric Encryption (The Handshake)**: Because sending a secret key over the open internet is dangerous, SSH initially uses a public/private key pair. The client and server use this complex math to securely negotiate and agree upon a single shared secret key without actually transmitting that key directly.
- **Symmetric Encryption (The Tunnel)**: Once the handshake is complete, the client and server use the single shared secret key they just generated to rapidly encrypt and decrypt all the data sent during the session. If someone intercepts the traffic, all they see is mathematical gibberish.
- **Hashing (Data Integrity)**: SSH attaches a cryptographic hash (like a digital fingerprint) to every packet of data. This ensures that the data has not been tampered with or corrupted in transit.
