```python
# indexing and slicing in numpy

import numpy as np
```

## Array Slicing

```python
# we start by indexing a 1d array first.

arr_1 = np.arange(1, 11)
print(arr_1)
```

### 1D Array Slicing

```python
# so, we have a simple 1d array, which is similar to a list in base python. Now, slicing a 1d array is very similar to how we do it in python, such as

print(arr_1[1:])
```

```python
print(arr_1[1:5])
print(arr_1[:5])
print(arr_1[2:])
print(arr_1[1:9:2])
print(arr_1[-3:9])
```

### 2D Array Slicing

```python
# so, for a 1d array, we were able to slice the array in a very similar manner, first we had result starting from index 1, then from ind 1 to 5, then from 0 to 5, then from 2, then 1 to 9, with a step of 2 and finally from -3 to 9th index.
```

```python
# now we move to a 2d array. For this, we must understand, that a 2d array consists of rows and columns. When we define a 2d array, we pass in a tuple or if we check the shape, a tuple is returned in this format > (row, column).

# with this understanding, it becomes very easy to extract/slice 2d arrays. below are some examples:

arr_2 = np.array([[1, 2, 3], [4, 5, 6]])

# now, this array has two rows, which are [1, 2, 3] and [4, 5, 6] and has 3 columns, which are [1, 4], [2, 5] & [3, 6].

# Now, to take out an element lets say, a single element, we simply point out to that element, using the same tuple format we discussed above. So, the first row has an index of 0 and the second row has the index of 1. Similarly, in each row, since we have three elements, their indexes are 0, 1 and 2. So, of lets say, we want to extract 6, so we are looking at row 1 and its element 2.

print(arr_2[1, 2])
```

```python
# as we can see above, it gives us that specific element. Now, if we want to extract a row, we simply pass in the row index.

print(arr_2[1])
```

```python
# and if we want to extract the column, we need to do (row, column) > but since we want to pass in an entire column, no row info must be passed. In this case, we use (:, column) as in the example below

print(arr_2[:, 2])
```

## Array Sorting

### 1D Array Sorting

```python
# so, 3 and 6 gets passed, which is the column at index 2 in the give array.

# next we move ahead with sorting in arrays. Now, sorting a 1d is simple, the numbers get sorted in ascending order. here is an example below:

arr_3 = np.array([1, 3, 5, 7, 8, 10, 3, 4, 8, 11, 21])
print(np.sort(arr_3))
```

### 2D Array Sorting

```python
# as we can see, we have the result in ascending order. And we use np.sort to do this sorting. Now, our data will never be a single list or a 1d array. Our data will be usually large tables, mainly 2d type of data. In that case, we will have to sort data in columns. We can also do that in rows. Let us create a 2d array and look at how sort works in case of 2d arrays

arr_4 = np.array([[3, 2, 1], [4, 5, 6], [8, 7, 9], [11, 14, 3]])
```

```python
# as we can see, we have this 2d array. Now, to sort this we need to specify if we need to sort the rows or the columns. lets first try sorting the rows

print(np.sort(arr_4))
```

```python
# here, we have did not pass any extra parameter, and if we see, each of the row got sorted in ascending order. Now, lets try this

print(np.sort(arr_4, 1))
```

```python
# we passed 1 alongside the array. Now, this is the axis value, where row is sorted when we pass 1 and column is sorted when we pass 0. Let's see what happens when we pass 0.

print(np.sort(arr_4, 0))
```

```python
# here if we focus on the last column, in the original array we had > [1, 6, 9, 3] > and in the one with axis as 0, we have 1, 3, 6 and 9. The column got sorted as per ascending order. Let us look at another example below

arr_5 = np.array([[3, 3, 3], [4, 4, 4], [1, 1, 1], [21, 21, 21], [10, 10, 10]])
```

```python
print(np.sort(arr_5, 0))
```

```python
# since all the rows were same, it looks like the rows got sorted, but in truth, the columns have been sorted.
```

## Filters in NumPy

```python
# it is also possible to use filters in numpy.

# Now, applying a filter is very unique to numpy. let us have a look at the below 1d operation

arr_6 = np.array([1, 2, 3, 4, 8, 9, 10, 789, 8984, 98984, 9869, 83])

# now, lets say, we only want even numbers from these and save those in a variable, we simply do the following

even_nums = arr_6[arr_6 % 2 == 0]
print(even_nums)
```

```python
# similarly we can have odd numbers

odd_nums = arr_6[arr_6 % 2 != 0]
print(odd_nums)
```

```python
# or lets say of we want to only show numbers below 10

print(arr_6[arr_6 < 10])
```

```python
# this was we can apply filters in 1d arrays. The expression, that we are using to filter out elements is called mask and it can be saved to a variable also, for example

less_than_ten_mark = arr_6 < 10

print(arr_6[less_than_ten_mark])
```
