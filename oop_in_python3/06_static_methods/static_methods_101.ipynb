```python
# static methods:

# instance methods use self, while class methods use cls. However, if we need basic utility functions, and do not need access to either the instance or class. Such methods are called static methods. 

# we decorate these methods using @staticmethod and as mentioned above, we do not use either self or cls. Let's look at an example below:

class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b
```

```python
print(MathUtils.add(3, 5))
print(MathUtils.multiply(3, 5))
```

```python
# while in the above example, we have not used init or other class / instance elements, these can be used with them also. Here is an example of that:

class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    @staticmethod
    def is_classic(year):
        """this method is used to simply check if a car is classic or not"""
        current_year = 2026
        return (current_year - year) > 25
```

```python
Car.is_classic(1999)
```

```python
Car.is_classic(2025)
```
