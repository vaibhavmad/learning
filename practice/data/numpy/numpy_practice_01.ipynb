```python
# C3. NumPy problem (home: Row 31). scores is a (4, 5) array: 4 learners × 5 modules, values 0–100. ids = np.array([101, 102, 103, 104]). With no loops:

# (a) for each learner, the index of their best module;
# (b) each learner’s scores rescaled to 0–1 within their own row: (x − row min) / (row max − row min);
# (c) the IDs of learners whose average is ≥ 70.
```

```python
import numpy as np
```

```python
# first lets create the array, we have scores with scores between 0 to 100, we create a randint array

scores = np.random.randint(0, 101, (4, 5))
print(scores)
```

```python
ids = np.array([101, 102, 103, 104])
# now, ids are given for each of the learners, we need to concat these ids to scores, to make it a complete array

student_scores = np.concat((ids, scores), axis=0)
print(student_scores)
```

```python
# nope, we need to hstack these together
student_scores = np.hstack((ids.T, scores))
```

```python
# okay, lets dry to reshape ids first, to match the dimensions
ids = ids.reshape(4, 1)
```

```python
student_scores = np.hstack((ids, scores))
print(student_scores)
```

```python
# now, lets come to the problems
# for each learner, the index of their best module

# best module is the max score, to be checked for each id. we print the first column and we then print the best score

print(student_scores[:, 0:1])
print(np.max(student_scores[:, 1:], 1))
```
