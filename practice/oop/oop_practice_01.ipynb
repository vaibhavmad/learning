```python
# okay, all works well till here. now, we need to write a static method, that checks if code is correct or not.


class Course:
    MAX_SEATS = 30
    courses_created = 0

    def __init__(self, code, title):
        self.code = code
        self.title = title
        Course.courses_created += 1

    @classmethod
    def from_string(cls, input):
        code, title = input.split("-")
        print(code, title)
        return cls(code, title)

    @staticmethod
    def is_valid_code(code):
        if code[0:2].isalpha() and code[-3:].isdigit():
            return "Valid Code"
        else:
            return "Invalid Code"


# checking this method
print(Course.is_valid_code("PY101"))
print(Course.is_valid_code("PY10"))
```
