```python
import numpy as np
```

```python
# let us start by creating a couple of arrays

arr1 = np.arange(11)
arr2 = np.random.random((2, 3))
print(arr1)
print(arr2)
```

```python
# okay, so now we have two arrays. numpy gives us the option to save these arrays in the form of files. the extension of these files is npy. it is very simple to save these by using the below syntax

np.save("arr1.npy", arr1)
np.save("arr2.npy", arr2)
```

```python
# now, on running the above commands, two new files were added to this folder. these are arr1.npy and arr2.npy.

# now, once the array information is saved in an npy file, we can load the array information and use it also.

load_arr1 = np.load("arr1.npy")
print(load_arr1)
```

```python
# as we can see, we were able to load and print the value of arr1 from the file.
# similarly

load_arr2 = np.load("arr2.npy")
print(load_arr2)
```

```python
# now images are made of pixels and the pixel information is stored in the form of arrays. we shall see the following example, by loading an image using matplotlib(copying the used commands as the usage and explanation is not given in the video).
#
# Also, for this one, we have added a npy file to the folder.
import matplotlib.pyplot as plt
```

```python
try:
    logo = np.load("numpy-logo.npy")

    # below commands are copied as is from the course
    plt.figure(figsize=(10, 5))
    plt.subplot(121)
    plt.imshow(logo)
    plt.title("NumPy Logo")
    plt.grid(False)

except FileNotFoundError:
    print("File missing!")
```

```python
# so this file, is a also made up of arrays. We can see that by simply printing it below.
print(logo)
```

```python
# now it is also possible to invert the colors of the image, by simply subtracting the file from 1.

dark_logo = 1 - logo
plt.subplot(122)
plt.imshow(dark_logo)
plt.title("NumPy Logo")
plt.grid(False)
```

```python
# as we can see we have now an image with inverted colors.
```
