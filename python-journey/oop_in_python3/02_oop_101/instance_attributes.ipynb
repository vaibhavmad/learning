```python
# We understand what an instance attribute is and 
# how to set, use and redefine instance attributes
```

```python
# An instance attribute is a variable and is used to store information regarding the instances of that class. Now, there are a few ways to set an attribute. For example:

class Person:
    pass

person_a = Person()
person_a.name = "John"

# while pylance here is giving us an error, however, if we run

print(dir(person_a))
```

```python
# in the above directory list, we can see 'name' has now been added

# also, if we run
print(person_a.name)
```

```python
# we can see the name getting printed out. also, this is the only attribute that we have created, if we try to call something else, like

print(person_a.names)
```

```python
# we can see, that we get an error. we can as above, set more attributes also, for example

person_a.age = 30
print(person_a.name, person_a.age)
```

```python
# both the attributes get printed out. Attributes can be redefined as well, meaning a new value can also be assigned to the attribute of an instance, just like a normal variable. for example:

print(person_a.name)
```

```python
person_a.name = "John Smith"
print(person_a.name)
```

```python
# there is another way to create instance attributes, which we can use while defining a class as below

class Bucket:
    height = 20
    width = 10
    depth = 18

# now, if we create a new instance of this class
new_bucket = Bucket()
# and then if we check it's directory
print(dir(new_bucket))
```

```python
# we can see above that depth, height and width have been added to the directory now. Also, since these attributes were defined while creating the class and values were passed, the same values of these attributes get passed over to any new instances that we create, such as

print(new_bucket.height, new_bucket.depth, new_bucket.width)
```

```python
# as we can see, the attribute values for this instance, got inherited from the main class itself. We can though redefine these or assign some other value to this as we did earlier, for example:

new_bucket.height = 40
```

```python
print(new_bucket.height)
```

```python
# while the other two attributes will remain same
print(new_bucket.height, new_bucket.depth, new_bucket.width)
```
