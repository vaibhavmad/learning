```python
# an instance method is a function, which is attached to a class. Basically meaning it can be only used by instance of the class. 

# While min, print, len, etc, are functions, whereas methods look a lil different, for example, append, it can not be used without a corresponding class.

# While methods are same as other functions, an instance method is always given 'self' as the first attribute. 

# The word self is purely a convention, we can use whatever word we want to from a design point, however, self is used primarily, so we shall be using the same too. 

# let us now understand the usage and significance of self attribute.

class Person:
    name = "John"
    age = 25

person_a = Person()
person_b = Person()

# now, both these instances will be similar, though stored in different memory references.

print(person_a.name)
print(person_a.age)

print(person_b.name)
print(person_b.age)
```

```python
# now, lets add a method to the above class(a function basically:) and we just add one attribute, self in this method

class Person:
    name = "John"
    age = 25

    def print_age(self):
        print("Called!")

person_a = Person()
person_b = Person()
```

```python
# lets now call this method
person_a.print_age()
person_b.print_age()
```

```python
# Now, we can see, that both the instances were able to call the method. Let's do one more slight modification here:

class Person:
    name = "John"
    age = 25

    def print_age(self):
        print("Called!", self) #we have just added self to the print value also

person_a = Person()
person_b = Person()
```

```python
# lets now call this method again
person_a.print_age()
person_b.print_age()
```

```python
# we can see from above, that while the static part of the print statement, (Called!) gets printed out, but when we print self, it is not same. We get exactly the following

print(person_a)
print(person_b)
```

```python
# so, the above experiment shows us, that we we use the keyword self and print it, and separately when we print the instances also, we get the same result. Since an instance is based on a class, when we try to call that instance, the use of word self inside the class, also refers to that instance. So, when we try to use something using person_a, and self is called, it refers to person_a, and same goes for person_b also.

# also, since the variables age and name were defined at the time of creation of the class, both the instances inherit that value from the class itself, as we saw above or as we can see below also:



print(person_a.name)
print(person_a.age)

print(person_b.name)
print(person_b.age)
```

```python
# now, however, if we change the age of person_a lets say

person_a.age = 50
# the age of person a will change, but the age of person b will remain the same as above
print(person_a.age)
print(person_b.age)
```

```python
# also, now if we do the following:

class Person:
    name = "John"
    age = 25

    def print_age(self):
        print("Called!", self.age) #we have added age to self.

person_a = Person()
person_b = Person()
```

```python
# and we do the following:
person_a.age = 50
```

```python
# the age of person a has again been changed. but if we run the method now
person_b.print_age()
person_a.print_age()
```

```python
# we can see, that while called! was printed, in the method, we had used self.age and when each of the instances called this method, their own value, that means person_a and person_b were passed on, self keyword was replaced and their individual values were printed. 
"""
so, when we did the following:

person_b.print_age()
person_a.print_age()

we send a message to python, that an instance of class Person() is called the instance method print_age() and the name of the instance is either person_a or person_b

next, python interpreter looks for this method in the class, and lets say if we have done person_a.print_age(), then the interpreter passes on the value: person_a and all its attributes want to call the print_age() method. Then, when the interpreter looks at the code, and wherever it sees self, it replaces the self with the instance name. so eventually, in the function, the following code runs:

print("Called!", person_a.age)

and we get output corresponding to person_a, which is 50 in this case.

"""
```

```python
# we can do further operations also by using the methods inside a class as follows:

class Person:
    name = "John"
    age = 25

    def set_age(self, new_age):
        self.age = new_age

person_a = Person()
person_b = Person()

# now, we have created two instances, lets check their ages
print(person_a.age)
print(person_b.age)
```

```python
# we can see, both carry the same age, however, now if we do the following:

person_a.set_age(50)
```

```python
# now, the following happened in this case

"""
when we ran this method, inter-p got to work and started looking for the class that person_a belongs to. It found the class, and then asked it, if there is a set_age method in this class. When class told it, yes, i do have one, it then asked the class how many attributes it needs?

The class told it, it needs only 1 attribute. It went back to the code that we ran and checked if 1 p is being passed. Since it was, it sent the following information to the class:
# hey class, your instance, person_a whose name is "John" and age is 25, wants to run the set_age method that belongs to you, with attribute value of 50.

Now, when the class got it, it passed the same info to the method, okay, here's the information, please process.

# Now, when the information was passed, self everywhere was replaced by the name of the instance, that is person_a. Finally, it processed the following info:

person_a.age = 50

Let's now validate this:
"""

print(person_a.age)
```

```python
# as we can see above, it is 50 while the age of person_b remains untouched.

print(person_b.age)
```

```python
# so, self helps us in accessing that instance and use it to run its methods and also change its attributes. 
```

```python
# lesson task

class Train:
    cars = 10
    car_capacity = 200

    def display_car_count(self):
        print(f"The train has {self.cars} cars.")

    def display_total_capacity(self):
        print(f"Total train capacity: {self.cars * self.car_capacity}")


train_1 = Train()

train_2 = Train()
train_2.cars = 5
train_2.car_capacity = 300

train_1.display_car_count()
train_1.display_total_capacity()

train_2.display_car_count()
train_2.display_total_capacity()
```
