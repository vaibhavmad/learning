```python
# a constructor's job is simple. It performs any initial setup for an instance. Let's say, we have a student class and we want to add its details to a text file or a db, as soon as the instance is created. This is used with the help of a constructor. Although, the main task of a constructor is to assign instance attributes and run any methods if needed.

# before we go ahead and go deeper in to constructors, there are couple of things that we need to understand
# 1. there are some methods, where we use double underscores as prefix and suffix to the name of the method. Such as __name__. Since, these methods use double underscore, either of the prefixes or suffixes is called dunder, where d is taken from double and under is taken from underscores. And such methods, where we use dunders are called dunder methods. 

# 2. When a new instance is created, python automatically calls a new method __new__, which when runs, allocates memory and creates the instance. This is followed by another __init__ method, which also runs automatically.Now, since __init__ is also called automatically, we use it to define our class variables and also run some functions if needed. 

# Now, in strict computer science terms, a method which allocates memory, in this case __new__ at the time of instance creation is called a constructor method. However, since this is hidden and also in general conversation, __init__ is also referred to as a constructor method. 

# we shall understand more with a few examples below

# lets say we have a class Person, lets create it, add an init method to it and then we will create the instance

class Person():
    def __init__(self):
        print("Init was called")

person_a = Person()
```

```python
# also, keeping this aside, one more point that we need to understand before we move on:

# when we create a class, lets say Dog, and we create a method in it, lets say bark. Now, if we create an instance, lets say romeo, and finally, when we run romeo.bark(), actually behind the scenes, it is being translated to > Dog.bark(romeo). Now, if we do not use self parameter while defining this method bark, and we try to run Dog.bark(romeo), we will get  TypeError stating that bark takes 0 arguments, but one was passed. This one argument that was passed was romeo. Only if we use self while declaring the method, the way we did it above while declaring __init__, then only we can run a method with an instance.

# Now moving ahead to our code. Here, we simply created an instance, but the print statement inside init was executed. This simply demonstrates that __init__ was called automatically when instance was created.

# now, we can create this __init__ with more parameters, such as 

class Person:
    def __init__(self, person_name, person_age):
        pass

# now, we know, if we create a new instance using class Person, the init method will be called. Lets try

person_a = Person()
```

```python
# now, there are two points to be noted here:
# 1. while we ran only Person() while creating the instance, the error states Person.__init__(), so this means, and not Person(). This means, that when any argument is passed in Person(), it means that is will be used as an argument in __init__() of class Person.
# 2. if we see the error above, it says, Person.__init__() missing 2 required positional arguments: 'person_name' and 'person_age'. So, as with any normal function, we need to pass on these two arguments, for this method to run properly.

# therefore we can do the following:

class Person:
    def __init__(self, person_name, person_age):
        pass


person_a = Person("John", 25)
```

```python
# now, while we were able to pass in this information in the form of arguments, these arguments did nothing, as it is like below

def random_function(num_1, num_2, num_3):
    print("Hello")

random_function(1, 2, 3)
```

```python
# in the above example, the information was passed, but it was never used. So was the case with the init method that we created. Now, this also means that we can use that information. We also remember, the following:

class Person():
    pass

person_a = Person()
person_a.age = 20
person_a.name = "John"

print(person_a.age)
print(person_a.name)
```

```python
# So, by using the above code, when we use an instance.variable, we can actually assign it a value, which becomes the value of an attribute that belongs to that class, but remains tied to that instance only.

# therefore, we can also do the following:

class Person:
    def __init__(self, person_name, person_age):
        self.name = person_name #similar to assigning person_a.name = "John"
        self.age = person_age #similar to assigning person.age = 25

# and we also know, when we create an instance, init runs automatically, and if the instance label is person_a, then we do the following to create this instance: person_a = Person(), but with two more parameters in init, we also need to pass two arguments, as we saw above. And also, self takes the value of the instance label. So, if we run the above, the following is happening:
"""
first, the instance is created in memory due to __new__
then __init__ runs
then, self is replaced by person_a
then person_a.name is assigned a value
and person_a.age is assigned a value, just like we did above after creating the instance
"""

# therefore, during instance creation only, we assign the value to its attributes like below:

person_a = Person("John", 30)
```

```python
# lets try to access these now
print(person_a.name)
print(person_a.age)
```

```python
# fun fact, when a new instance is created and then when we assign an attribute to it, it is store in a dict, which is tied to that instance. it can be accessed as below:

print(person_a.__dict__)
```

```python
# also, if we assign attributes at the time of class creation, for example

class Person:
    name = "John"
    age = 20

# and then create an instance, 
person_a = Person()

# now, if we try to check the attributes of this instance, lets see what we get
print(person_a.__dict__)
```

```python
# we get an empty dict, because the instance attribute is never created in this manner. class attributes are created. and if we do
print(Person.__dict__)
```

```python
# we can see above, we have name and age both defined for the class. therefore, till the time, a value is not assigned specifically to instance.attribute, it will not show up in its dict

# we can also act upon the attributes inside __init__, for example:

class Person:
    def __init__(self, person_name, person_age):
        self.name = person_name.title()
        self.age = person_age

person_a = Person("john", 30)

print(person_a.name, person_a.age)
```

```python
# we see here that john got converted to John at the time of instance creation itself.

# also, from a learning pov, I am using person_name, person_age, usually it is done as below:

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```
