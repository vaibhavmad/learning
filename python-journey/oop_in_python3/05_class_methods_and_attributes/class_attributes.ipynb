```python
# class attributes are attributes that are defined at class level. It is used mainly for having an attribute which is going to be common amongst the instances.

# For example, we design a Car class. All cars(well mostly) have 4 wheels. So, instead of setting it every time a new instance is formed, we can set it once while designing the class. Now, this attribute will belong to the class and each instance will be able to access it from the class.

# let us understand it with the below example:

class Car:
    wheels = 4

    def __init__(self, make, model):
        self.make = make
        self.model = model

car_a = Car("Maruti Suzuki", "Swift")
car_b = Car("Hyundai", "Creta")
```

```python
# now, if we access the dicts of these instances
print(car_a.__dict__)
print(car_b.__dict__)
```

```python
# we see that only make and model are visible, however, if we print the following:
print(car_a.wheels, car_b.wheels)
```

```python
# we see that 4 and 4, is printed out for each. Since, we defined earlier that this attribute belongs to the class and not the instance, if we print
print(Car.wheels)
```

```python
# we again get 4, which belongs to the Car class basically. However, on printing the following:
print(Car.make)
```

```python
# we get an error. Also, this is mutable, so if we do
Car.wheels = 5
print(car_b.wheels, car_a.wheels)
```

```python
# we get an updated number. Now, as we saw above, the class attribute is declared before the init method. So, even this gets executed at the time of creation of the instance. We can use this to a good use also. For example, keeping a count of total instances created

class Bike:
    wheels = 2
    count = 0
    def __init__(self, make, model):
        self.make = make
        self.model = model
        # and now, we can access count using class and increment it within init, so every time init runs or in other words, a new instance gets created, we increment the count
        Bike.count += 1
        # lets put this to use

bike_a = Bike("Hero", "Splendor")
bike_b = Bike("Bajaj", "Pulsar")
bike_c = Bike("Honda", "CBR")
```

```python
print(Bike.count)
```

```python
# as we can see here, we get 3 as the count in this case.
# Also, one important point to be noted here is that if we pass on a value to the same identifier as a class attribute under init, then it will over write the class attribute value. lets understand it, using the below example

class Dog:
    legs = 4

    def __init__(self, breed):
        self.breed = breed
        self.legs = 5

romeo = Dog("Doberman")
juliet = Dog("Shepherd")

print(romeo.legs, juliet.legs)
```

```python
# in both the cases, the attribute created in the instance with the same identifier or name as the class attribute, the instance value will take precedence. 
```
