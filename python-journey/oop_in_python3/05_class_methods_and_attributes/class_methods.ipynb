```python
# a class method is a method similar to an instance method, but instead of running in tandem with an instance, it is joined with a class. These methods are used to operate on the class attributes and also and has some other use cases also. We shall understand all of it, using examples. 

# Now, the point to be noted here is that class methods use a decorator > @classmethod, which tells the python behind the hood that the following method is to be called in tandem with a class. lets define a class and a class method first.

class Dog:
    legs = 4

    def __init__(self, breed: str):
        self.breed = breed

    @classmethod
    def change_leg_count(cls):
        print("class method was called")

# while we noted above, that we need to pass in a decorator -> @classmethod, we must also pass in a parameter, which can be anything, but we use cls. This is just like self, and is a self reference to the class. lets run this below.

Dog.change_leg_count()
```

```python
# as we can see, we were able to call this method, but referencing with the class. Now, this can also be referenced using an instance. When we pass in instance.class_method(), python in the background checks its type and then passes on the class, so instance.class_method() gets translated to type(instance).class_method(). Lets have a look at the example:
romeo = Dog("Shepherd")

romeo.change_leg_count()
```

```python
# Now, let us check the use case of a classmethod. Practically they serve 3 use cases. We cover each one by one below.

# 1. Alternative constructors > We use __init__ as the constructor, where when a new instance is created, the init function runs and passes on the instance attribute values given. This is very simple, for example:

class Dog:
    def __init__(self, legs, breed, name):
        self.legs = legs
        self.breed = breed
        self.name = name

# now when we do this:
new_dog = Dog(4, "Shi tzu", "Romeo")
# the init method is called, instance attributes are created and are stored in the dict, as we have seen earlier. We can call these using new_dog.legs or new_dog.name
```

```python
# however, consider, there is some data that we need to use to create an instance, lets say, such as 4-shi_tzu-romeo or lets say a database row. We can use class methods for that, below is an example. we keep the same above example and we now want to create an instance, using given string in this format: 4-shi_tzu-romeo

class Dog:
    def __init__(self, legs, breed, name):
        self.legs = legs
        self.breed = breed
        self.name = name

    @classmethod
    def from_string(cls, input_str: str):
        # now, we have an input string, this function is also taking a string. if we remove - from the string and separate the words, we can use that to run init again, to basically calling the same code, as we do while creating a new instance, as in below:
        legs, breed, name = input_str.split("-")
        return cls(legs, breed, name) #this is exactly like calling Dog(legs, breed, name)    
```

```python
# now, we have both the methods and we can use both to create an instance, lets have a look
dog_a = Dog(4, "shepherd", "juliet")
dog_b = Dog.from_string("4-shi_tzu-romeo")
```

```python
print(dog_a.__dict__)
print(dog_b.__dict__)
```

```python
# this is one of the use cases. Second use case is when you need to operate on a class attribute. Now, we might directly change its value, since it is mutable, but it goes against the principles of OOP. We shall understand each of these cases in the upcoming examples. 

# first, lets do a simple example

class Bike:
    wheels = 2

    def __init__(self, make, model):
        self.make = make
        self.model = model

    @classmethod
    def update_wheel_count(cls, new_count):
        if not isinstance(new_count, int):
                    raise TypeError("Wheel count must be a number")
        if new_count < 2:
            raise ValueError("Wheel count can not be less than 2.")
        

        cls.wheels = new_count


    def wheel_change_cost(self, per_tyre_cost):
        print(self.wheels * per_tyre_cost) 
```

```python
new_bike = Bike("Bajaj", "Pulsar")
new_bike.wheel_change_cost(10)
```

```python
# Now, since we had two wheels by def, we got the cost as 20. However, lets say now, we have started producing all bikes with a sidecar. Instead of doing Bike.wheels = 3, we can run the given function

Bike.update_wheel_count(3)
```

```python
# and now, if we run 
new_bike.wheel_change_cost(10)
```

```python
# our answer is 30. Again the question is, we can do Bike.wheels = 3 or even we can do it for that instance. But, a new dev working with this class, by mistake can update incorrect info, such as Bike.wheels = -2 or Bike.wheels = 0

# by using this class method, we have added the validations in place, such as

Bike.update_wheel_count(-2)
```

```python
# or
Bike.update_wheel_count("three")
```
