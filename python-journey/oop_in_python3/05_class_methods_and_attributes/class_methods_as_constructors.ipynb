```python
# while we have already covered this before, but we revise it here.

# we start by creating a simple class and then creating an instance

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

john = Person("John", 26)
```

```python
# now lets say, we have data, where instead of age, we have the birth_year of that person. and if we follow the same above process, we will end up assigning the birth year to age. for this, we create a class method. 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_birth_year(cls,name, birth_year):
        # we are keeping it simple for now
        current_year = 2026
        age = current_year - birth_year
        return cls(name, age)
```

```python
# lets try it out
sunny = Person.from_birth_year("Sunny", 1988)
```
