```python
# create an Employee class
# its attributes will be: company, bonus_amount, pay_rise_percent
# instance attributes: name, pay, role -> here, though it is not specified, we take the pay as annual pay

# class methods: change the bonus amount, change the pay rise percent
# instance methods: show pay with bonus, increase annual pay
```

```python
# so, we start by creating the class
class Employee:
    
    # now we create class attributes
    company = "Python Inc"
    bonus_amount = 10_000
    pay_rise_percent = 10
    
    # now, lets move ahead with __init__ and create instance attributes
    def __init__(self, name, pay, role):
        self.name = name.title()
        self.pay = pay
        self.role = role

    # now, we move ahead with class methods
    @classmethod
    def change_bonus(cls, new_amount):
        # now, bonus amount needs to be a number and it can not be less than 0, we put a validation for this
        if not isinstance(new_amount, (int, float)):
            raise TypeError("bonus amount must be a number")
        if new_amount < 0:
            raise ValueError("bonus amount can not be negative")
        # if both these pass, then we can assign the value of new_amount to bonus_amount
        cls.bonus_amount = new_amount
```

```python
# lets validate this much, then we copy the above code and keep working on it.
emp_a = Employee("Ajay", 60000, "Manager")
```

```python
# great, this works also. We copy the entire class and paste it, to keep working on it.
# so, we start by creating the class
class Employee:
    
    # now we create class attributes
    company = "Python Inc"
    bonus_amount = 10_000
    pay_rise_percent = 10
    
    # now, lets move ahead with __init__ and create instance attributes
    def __init__(self, name, pay, role):
        self.name = name.title()
        self.pay = pay
        self.role = role

    # now, we move ahead with class methods
    @classmethod
    def change_bonus(cls, new_amount) -> None:
        # now, bonus amount needs to be a number and it can not be less than 0, we put a validation for this
        if not isinstance(new_amount, (int, float)):
            raise TypeError("bonus amount must be a number")
        if new_amount < 0:
            raise ValueError("bonus amount can not be negative")
        # if both these pass, then we can assign the value of new_amount to bonus_amount
        cls.bonus_amount = new_amount

    # next we work on change_pay_rise_percent function. again, both the above validations work on it also. it can not be negative and it needs to be a number
    @classmethod
    def change_pay_rise_percent_to(cls, new_percent) -> None:
        if not isinstance(new_percent, (int, float)):
            raise TypeError("Pay rise percent must be a number")
        if new_percent < 0:
            raise ValueError("Pay rise percent can not be negative")

        cls.pay_rise_percent = new_percent

    # now, lets add the instance methods
    def show_pay_with_bonus(self) -> None:
        print(f"{self.name}'s annual pay with bonus is: {self.pay + Employee.bonus_amount}")

    def increase_annual_pay(self):
        self.pay = round((self.pay + (self.pay * Employee.pay_rise_percent/100)), 2)
```

```python
# lets now test all of it.

emp_1 = Employee("ajay", 50000, "Manager")
emp_2 = Employee("vijay", 60000, "Senior Manager")
emp_3 = Employee("sujay", 70000, "Director")
```

```python
print(emp_1.__dict__)
print(emp_2.__dict__)
print(emp_3.__dict__)
```

```python
print(emp_1.company, emp_1.bonus_amount, emp_1.pay_rise_percent)
print(emp_2.company, emp_2.bonus_amount, emp_2.pay_rise_percent)
print(emp_2.company, emp_2.bonus_amount, emp_2.pay_rise_percent)
```

```python
# cool, these work fine. Now, we change the bonus amount and then we change the pay rise percent
Employee.change_bonus(15000)
Employee.change_pay_rise_percent_to(15)
```

```python
# now, lets check their pay with bonus
print(emp_1.show_pay_with_bonus())
print(emp_2.show_pay_with_bonus())
print(emp_3.show_pay_with_bonus())
```

```python
# now, lets increase their annual pays
emp_1.increase_annual_pay()
emp_2.increase_annual_pay()
emp_3.increase_annual_pay()
```
