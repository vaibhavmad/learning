```python
# task 3 -> Define  a new class method to accept a new brand and set the attribute. It must only be letters and not longer than 10 characters.

class Laptop:
    brand = "Apple"
    def __init__(self, ram: str, processor: str) -> None:
        self.ram = ram
        self.processor = processor

    @classmethod
    def change_brand_to(cls, new_brand: str) -> None:
        # also, it must be a string in addition to the two givens
        if not isinstance(new_brand, str):
            raise TypeError("Brand name must be a string only.")
        if not new_brand.isalpha():
            raise ValueError("Brand name must only contain characters.")
        if len(new_brand) > 10:
            raise ValueError("Brand name must be less than 10 characters.")

        cls.brand = new_brand
```

```python
# lets put above to validation
Laptop.change_brand_to("dell09")
```
