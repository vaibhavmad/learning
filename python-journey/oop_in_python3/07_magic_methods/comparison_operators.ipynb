```python
# comparison operators -> dunders that work behind the scenes are:

"""
== __eq__ -> equal
!= __ne__ -> not equal
>  __gt__ -> greater than
<  __lt__ -> less than
>= __ge__ -> greater than or equal to
<= __le__ -> less than or equal to

"""
```

```python
# Now, like __str__, when ever we execute one of the above operations, the respective operator, starts looking for its respective dunder in the directory of that object. Now, like we were able to manipulate what would be printed, we can also manipulate what is being compared. Let us understand this using the below example:

class House:
    def __init__(self, rooms):
        self.rooms = rooms

# we have creates a simple class. Lets create a couple of instances of this.
```

```python
house_a = House(5)
house_b = House(5)
```

```python
# Now we have a couple of instances, let us compare them for equality
print(house_a == house_b)
```

```python
# this returns as false. When we check house_a == house_b, we are actually running house_a.__eq__(house_b). Now, two instances even with the same attributes as above, are not the same objects. Therefore, we get false when we compare them. But, we can change this behavior when we define our own __eq__ and set up what is being compared. In the above example, actually their memory location was being compared, which is not the same. Now, let us look at the below example:

class House:
    def __init__(self, rooms):
        self.rooms = rooms

    # now instead of comparing the default, we create a custom __eq__, where we compare the value of the attributes. 
    # also, since two objects are being compared, we need to also pass in two parameters, self and other.
    # self gets assigned to the object on the left, while other gets assigned to the object on the right
    def __eq__(self, other):
        return self.rooms == other.rooms
```

```python
# here, we have recreated the eq, so, not when we run the comparison between both the objects, the value of their attributes will be compared.

house_a = House(5)
house_b = House(5)

# we remember that when these were compared earlier, the result was false. Let us see now

print(house_a == house_b)
```

```python
# now, we can see that the result comes out to be true. Similarly for our class, we can create custom operations based on the operator is being called by defining methods for each of the dunders.
```
