```python
# magic methods, also called as dunder methods or special methods. before we go further, lets run the below code:

print(f"List: {dir(list)}\n")
print(f"Dict: {dir(dict)}\n")
print(f"String: {dir(str)}\n")
print(f"Integer: {dir(int)}\n")
```

```python
# as we can see from above, the directory of the data types that we just printed, gave us the methods for that data type. Now, while for list: append', 'clear', 'copy', 'count', 'extend', 'index', 'insert', 'pop', 'remove', 'reverse', 'sort' and for lets say Int: as_integer_ratio', 'bit_count', 'bit_length', 'conjugate', 'denominator', 'from_bytes', 'imag', 'is_integer', 'numerator', 'real', 'to_bytes' > these are the normal methods. 

# However, when we look at the entire directory, we can also see some methods in the format > __methodName__. These are called dunder methods, special methods or magic methods.

```

```python
# These methods are called behind the scene by a range of every day Python activities such as print, len etc. For example, len function calls __len__ which is present in the dir's above also. So, when we call len on an object and check if this object has __len__ method, __len__ is then called. Therefore we can say that len is like an interface which calls __len__ in the background. 

# let us see this with an example below:

new_list = [1, 2, 3,]
print(len(new_list))
```

```python
# and now, when we use __len__(), let's see what we get
print(new_list.__len__())
```

```python
# similarly when we look at the magic methods of an obj, lets say list, when we call print > __str__ is called, when we call == __eq__ is called and so on. 
 
# Also, couple of other points regarding python:
# 1. Python is a set of rules, however, the program that we use in our machine is called CPython. Now what does this mean? Well, let us look at how python is executed > Our python code, that we write is converted into bytecode by the Python Compiler, and this is then passed on to the Python Virtual Machine, now, this virtual machine is written in C language. It parses the bytecode instructions and executes them one by one. 

# While, the one we are using is CPython, there are others also, like PyPy, Jython, MicroPython etc. 

# 2. In CPython, all the functions, rather everything that gets executed in python is actually written in C language. When we execute any python code, it gets converted to bytecode by the compiler. Once this is done, the Python Virtual Machine which is written using C language also, processes the bytecode in C language, in operations that are to be performed also written in C and returns the python based response to the user. So, in the backend, everything is being done in C. For that matter, even __len__ special method gets executed in C. Also, this compiler + the PVM(python virtual machine) together is called the python interpreter. So, even though compilation is happening behind the scenes, but at run time, it is being interpreted and compiled at the same time.

# Now, moving back to dunder methods. If we look at these methods, they are created in this format: __method_name__. Now, this is an effort by the creators to keep normal vocab open to be used by the developers. If there were no dunder methods, then the list of methods not meant to be named or we can call reserved keywords, that we can not use to create variables would have been huge. To prevent this from happening, they used ____ dunders.
```

```python
# Now, how a __dunder__ will behave totally depends upon what it is being called on. For example, we know that when we use + or i should say __add__ on a couple of strings > it concatenates the strings, however, when we use these on two numbers(int or float) it gives us the sum of those values. 

# therefore, __add__ is not something that defines the operation. it is a method of that specific object and that object defines what __add__ can do to it. So, when we use +, in the backend, it is checked if both the objects have __add__ method. if yes, the __add__ operation for that specific object is called. This is the reason, we can not add a string to an int in python.
```
