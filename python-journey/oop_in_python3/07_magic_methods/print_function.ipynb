```python
# in this lesson, we will see what happens when we use print function. As we saw last, a function will look for a specific dunder method in the dir of the object it is being called upon. Once, it is located, it will be executed on the basis of how that dunder is supposed to act with that object. 

# now, as per cpython documentation, when we call a specific function, a dunder tied to it is called. For == as we mentioned above, __eq__ is called, for + __add__ and similarly for print, the following functions are called -> __str__, __repr__. Now, how can two functions be called for one command. Let us understand it from the below example:

# for this, we create a simple class and give it a couple of attributes. then we create an instance for it.

class Ship:
    def __init__(self, crew, length):
        self.crew = crew
        self.length = length

ship = Ship(10, 25)
# now, this is a ship with crew of 10 and length of 25. Let's try printing this instance.

print(ship)
```

```python
# Now, when we print this instance, we get -> <__main__.Ship object at 0x107cce4b0>, means this is an object of type Ship, store at reference 0x107cce4b0 in memory.

# however, let us look at the below code now:
class Ship:
    def __init__(self, crew, length):
        self.crew = crew
        self.length = length

    def __repr__(self):
        return "repr was called"

ship = Ship(10, 25)
```

```python
# now lets try printing ship
print(ship)
```

```python
# cool, lets try something on top of this now.

class Ship:
    def __init__(self, crew, length):
        self.crew = crew
        self.length = length

    def __repr__(self):
        return "repr was called"

    def __str__(self):
        return "str was called"

ship = Ship(10, 25)
```

```python
print(ship)
```

```python
# so, we can clearly observe that when there was neither __str__ or __repr__ in the function, we got the details of the type and memory location. When we added __repr__, it got printed and then when we added __str__, that got printed. So, in terms of what is to be printed, if an object has __str__, that get's first priority, if it does not has a str, but repr, then repr gets printed. If there is nothing then the default memory reference get's printed. 

# Now, for a class, let us understand what is the use of __repr__ and __str__. __str__ is used to print out a human friendly representation or details of an object, while repr is used to print out dev friendly details of an object.

# There is no format in terms of what to output from str or repr, but we must always return a string. Let's finalise both repr and str for Ship class.


class Ship:
    def __init__(self, crew, length):
        self.crew = crew
        self.length = length

    def __repr__(self):
        return f"Ship({self.crew}, {self.length})"

    def __str__(self):
        return f"Ship with {self.crew} crew count and {self.length}ft length."

ship = Ship(10, 25)
```

```python
print(ship)
print(ship.__repr__())

# worked around a bit to print how repr functions, since, print alone wont print it, due to the presence of str in the class.
```

```python
# Now, there is one exception to the priority of str vs repr. It is, when we try to print a list of objects. Then repr is printed out. Here's an example:

ship_b = Ship(30, 30)
```

```python
print([ship, ship_b])
```

```python
# as we can see from above, even though str is present, but repr got printed when we printed both the objects enclosed in a list.
```
