```python
# in this lesson, we learn about class and instance.

# What is a class?
# In simple terms, it is a template that describes everything about the things we can make with the template. For example, if I am making a bucket, I will be using a special type of metal cast die, on which the buckets will be modeled. This cast die is the class. And for different types of buckets, we will be using different cast dies. So, every cast die will contain information for what type of bucket it can be used to manufacture, such as height, width and depth etc.

# What is an instance?
# These are the items/things that we make using classes. Continuing from the above example, a bucket made using the cast die class will be its instance. 

# Now, to declare a class, we use class keyword, as below

class Person:
    pass

# note that the first alphabet of the name of the class is always capital and if we want to by any chance create a class using two words, then it should be like this: HumanPerson. Now, once we use class keyword and give it a name like we did above, we are telling python that we are creating a new template.

# For now, this template is empty, but we can still use it to create an instance, like below

person_a = Person()
person_b = Person()

# now, these are two separate instances of the class Person and are not connected to each other. If we simply print these, we get

print(person_a)
print(person_b)
```

```python
# as we can see above, we get two memory references, one for each of the instances. we can also do

print(id(person_a))
print(id(person_b))
```

```python
# and as we can see they have different IDs in python. 

# Now, most classes have the following:
# 1. Attributes: information for each instance
# 2. Methods: action that each instance can do
```
