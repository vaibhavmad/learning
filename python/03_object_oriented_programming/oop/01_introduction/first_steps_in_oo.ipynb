### Understanding OOP in Python

If we take a basic list in python, there are methods that we can use to work on a list. For example, index, len, sort etc. These methods were created when python was created. However, some of these methods won't work on an integer. Now, list and int are objects and they have their own methods or behaviors.

Same is the case for other similar data types, or even libraries. Like if we take date time, the methods of datetime wont work with lists or int. 

This is the basic crux of OOP. We define a thing(object) and it can do some actions(behaviors or methods) and can store data(variables). We shall cover these with code examples also. 

```python
# we start by running a simple code below
my_list = [1, 2, 3]
print(dir(my_list))
```

```python
# Now as we can see above, we have used a dir function, which stands for directory. For now, if we ignore the dunder functions, we have the following list:

# 'append', 'clear', 'copy', 'count', 'extend', 'index', 'insert', 'pop', 'remove', 'reverse', 'sort'

# now, these are some functions that we can use on lists. For example
my_list.append(4)
print(my_list)
```

```python
# Now we can see above, we just used a method append on the list object. However, if we now turn the list into a tuple and try doing the same again, lets see what do we get

my_list = (1, 2, 3, 4)
my_list.append(5)
print(my_list)
```

```python
# as we can see we get AttributeError: 'tuple' object has no attribute 'append'. This means, the action of appending something does not exist for a tuple. And, now if we run a dir on my_list again, we get
print(dir(my_list))
```

```python
# Now, we get count and index appear here also. So, both lists and tuples have these in common and they will work on both. and if we run
my_list.count(1)
```
