```python
# task 1
# Create a computer class with a constructor, with attributes: ram(int), platform(str), processor(str)
# use type hints
```

```python
class Computer:
    def __init__(self, ram: int, platform: str, processor: str):
        self.ram = ram
        self.platform = platform
        self.processor = processor
```

```python
new_computer = Computer(2048, "Mac", "M1")

print(new_computer.__dict__)
```

```python
# task 2
# make a climber class, attributes: name(str) - title case, height(int), define a list above the class and append the climbers name to it.
```

```python
climber_name = []

class Climber:
    def __init__(self, name: str, height: int):
        self.name = name.title()
        self.height = height
        climber_name.append(self.name)
```

```python
new_climber = Climber("Ankit", 160)
```

```python
print(new_climber.__dict__)
print(climber_name)
```

```python
new_climber2 = Climber("vaibhav", 160)
```

```python
print(new_climber2.__dict__)
print(climber_name)
```
