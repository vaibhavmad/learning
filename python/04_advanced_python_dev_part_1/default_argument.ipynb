```python
# let us look at the below code

account_balance = {"savings": 200000.00, "current": 100000.00}

# the above dict simply has two account name keys and balance associated with these
# we write the below function, to add balance to the accounts.


def add_balance(amount: float, account: str) -> float:
    """Adds given amount to account type and returns the updated amount in account type."""
    account_balance[account] += amount
    return account_balance[account]


# now, let us run this once.
add_balance(100000, "savings")
```

```python
print(account_balance)
```

```python
# so, the function works fine. Now, however, let us assume, that 95% of our transactions happen in savings account, in that case, there should be a way, using which we only select an account, if current account transaction happens, and otherwise savings remains the default account type. We use default arguments for this. Below is the example:


def add_balance(amount: float, account: str = "savings") -> float:
    """Adds given amount to account type and returns the updated amount in account type."""
    account_balance[account] += amount
    return account_balance[account]


# in the above code, in front of account now we have added 'savings'. Let us now try adding money to the accounts. However, we do not mention the account type.

add_balance(25000)
```

```python
# as we can see, 300000 was the balance in savings and on adding 25000, without specifying the account type, money got added to savings. However, if we need to add money to current, then we will have to mention the account name.

add_balance(25000, "current")
```

```python
print(account_balance)
```

```python
# This is what default arguments are used for. Only point to be noted here is, that when we create a function with def arguments, they should follow any non default parameters, like below

# def add_balance(amount: float, account: str = "savings") -> float: (CORRECT)
# def add_balance(account: str = "savings", amount: float) -> float: (INCORRECT)
```
