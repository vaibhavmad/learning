# Mutability in Python

```python
# data once created if can be changed, it is considered mutable and if it can not be changed, it is considered immutable.

# for example, lets say if we create the below dict in python

friends = {"Jon": 1, "Ron": 2, "Amen": 3}

# here, data is the dictionary and not the variable i.e friends. When we create one such variable, we are simply creating a dict in the memory and giving it a name of friends.

# now, when we do

friends_dict = friends
# both these are pointing to the same dictionary that we had created above.

# we know that it is possible for us to take out the id of an object in python. this id is the address of the memory where that object is stored. If we print the ids of the above two, we should get the same ids.
print(id(friends))
print(id(friends_dict))
```

```python
# as we can see that the IDs of the above two objects is same.

# now lets say if we create another dict, with the same name and same information, let's see what id is given to it.

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))
```

```python
# while now we had the same variable name and this variable was pointing to the same type of data as well, their ids were different. This is because, python program runs in a sequential manner.

# when we create the first variable friends, basically we are creating a dict with the information of the friends and storing it in memory. Then python gives it a name of friends. Now, we must understand that there is a memory id 4459509952 and it has been allocated to a dict, or in other words it is not available or is occupied now.

# then we move ahead and run the same command again, where another dict even though with the same values, is created and stored in the memory. then a same name is also given to it. Now, however, when we print its id, it is different from the previous one. this is because these are two different objects, located at two different memory locations.

# now lets use the same dict again

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))

# now, we do the following
friends["Amen"] = 10
print(id(friends))
```

```python
# now, when we executed friends['Amen'] = 10, we did not make a new object. we simply modified the existing dict or in other words, we mutated the existing dict. hence, we can also say that a dictionary in python is a mutable data type or it can be modified after it is created.

# there are 4 data types which are not mutable in python, or data types, which once created can not be edited. any edit to these data types will return in creation of a new data altogether. these are -> int, float, strings and tuples. let us understand more using below examples:

int_1 = 1
# we have created int data type and stored it in memory, gave it the name of int_1. Now, lets check it's id
print(id(int_1))
```

```python
# now if we make any changes to the int_1, lets say, by adding 1 more to it
int_1 = int_1 + 1
# now, let us check its id.
print(id(int_1))
```

```python
# as we can see, when 1 became 2, since we had added 1 to it, it got stored at a new memory location, since, the original 1 could not be changed. same, as we mentioned is the case with strings, floats and tuples. So, when ever we make any changes to an existing int, str, float or a tuple, a new item of the same data type gets generated and is stored at a diff place in the memory. it happens, since these data types are immutable.

# now, since we are aware of what is mutability, let us understand why it is dangerous or what all do we need to take care of when working in python to prevent bugs related to mutability.

# in python, now since we know that variables are not buckets that contain data, but labels, which point to data or an object which is present in the memory.

# now, multiple variable names or labels can point to the same object in memory. If we change the data using one label, the data which other labels point to also changes. This happens with data which is mutable in nature, i.e other than int, str, float and tuples.

# let us now look at most common ways mutability causes bugs in python.

# 1. Aliasing
# suppose we have a list of numbers assigned to variable numb1. we then assign numb1 values to numb2. Now, both numb1 and numb2 are pointing to the same list, like below:
numb1 = [1, 2, 3]
numb2 = numb1

print(id(numb1))
print(id(numb2))
```

```python
# as we can see, both these variables point to the same object in memory. Now, if we make changes to the data using numb1, the same changes will reflect in numb1 also.

numb2.append(10)
```

```python
# 2. Shared Class Attribute
# let us look at a few examples below to understand, where this becomes a problem


class Team:
    members = []

    def __init__(self, name):
        self.name = name

    def add(self, person):
        self.members.append(person)


# as we can see above, we have one class, Team. In this class, we have a list members, which is a class attribute. Now, when we create an new instance of this class such as
india = Team("india")
pak = Team("pak")
# by executing above, we have created two instances of team class. Now, however, if we start by adding a few members to both teams, we get
india.add("virat")
india.add("sachin")
india.add("vaibhav")

pak.add("afridi")
pak.add("wasim")
pak.add("akram")
```

```python
# now, let us check the lists for both
print(india.members)
```

```python
# as we can see, even though, we had referred to the list using the two instances, since, this list was shared between both the instances, the same list was being called and being updated.

# same is the case with using a mutable default argument in a class's __init__ method, every instance of that class that uses the default will share the exact same object.

# 3. Mutable Default Arguments (The Classic Trap)
class Player:
    def __init__(self, name, inventory=[]):
        self.name = name
        self.inventory = inventory


# Create two separate players
player1 = Player("Alice")
player2 = Player("Bob")

# Alice finds a sword
player1.inventory.append("Sword")

# Bob magically has the sword too!
print(player2.inventory)
# Output: ['Sword']
```

```python
# the solution is either using None instead of [] or using a rebind version, by declaring an instance list after instance creation.

# Another issue is
# 4. Shallow vs deep copy:
# In case a list has inner lists, using .copy as we used above, only creates a new outer list, but inner lists are still shared. Below is an example of the same:
grid = [[0, 0], [0, 0]]
g2 = grid.copy()  # only a new outer list has been copied, both the outer lists, grid and g2 still share the same inner lists. Here, we make a change to g2's inner list.
g2[0][0] = 9
print(
    grid
)  # [[9, 0], [0, 0]]  ← inner lists are still shared, and the changes now reflects in the original list.
# This can be improved by using copy.deepcopy as we use it below
import copy

g3 = copy.deepcopy(grid)  # copies everything, inner lists included
g3[0][0] = 15
print(grid)  # grid is still the same
print(g3)
```

```python
# Another important factor is
# 5. is vs ==. This follows straight on from your id() work:

a = [1, 2]
b = [1, 2]
a == b  # True  — same values
a is b  # False — two different objects (different id)

# == checks for same values, while is tells us if both the variable names are pointing to the same data or not.

# Use is only for None (if x is None:). That’s the check our default-argument fix relies on.
```

```python
# 6. += behaves differently on mutable and immutable types:

# let us look at the below examples:
j = [1, 2]
y = j  # now both y and j point to the same list
print(y)
y += [
    10
]  # here we are asking the existing list to add an element 10 to itself. so, still the list remains the same
print(y)
print(j)

j = (
    j + [29, 30]
)  # here however, we have first add 29, 30 to the existing list, now this list is [1, 2, 10, 29, 30], while another list [1, 2, 10] still exists and y points to it. while after creating this new list, we assign the variable j to it.
print(j)
print(y)
```

```python
# next we move to argument mutability in python, for this, we consider the below dictionary

friends = {"jon": 21, "mon": 22, "gon": 23}
# it is basically a list of friends and when we last saw them
```

```python
# now, we write a function below


# def see_friends(friend_list, friend_name):
#     friend_list[friend_name] = 0


# this function updates the last saw time to 0.
# before we move ahead, let us check the id of friends and each of the keys
print(id(friends))
print(id(friends["jon"]))
print(id(friends["mon"]))
print(id(friends["gon"]))
```

```python
# now, we make a slight change to the function, we also print the id of the list which is being passed through in the function itself. next up, we print the id, run the function and run the id again
def see_friends(friend_list, friend_name):
    print(id(friend_list))
    friend_list[friend_name] = 0


print(id(friends))
see_friends(friends, "jon")
print(id(friends))
```

```python
# even though the id inside also changed, but it remained as is.

# here's another example of how mutability impacts arguments

age = 20


def increase_age(user_age):
    user_age = user_age + 1


print(id(age))
increase_age(age)
print(id(age))

# now, from the above example, we see that age is 20 and lets say has id a(we will check this later). Now, we run the function increase_age on age. Basis the code, we now assume that, we are running age = age + 1, and age becomes 21. As we saw earlier, when we change the value of an int, the variable gets a new id, so in our case, when we print the id of age again, it should be different from the one that was printed before running the function. Let's run the code.
```

```python
# here, this illusion is for the fact that, When we call the function, Python runs user_age = age, so both labels point to the same 20. Then user_age = user_age + 1 creates a new int object 21 and moves only user_age onto it. age still points to 20. When the function ends, user_age is discarded. So age stays 20, with the same id.
print(age)
```

```python
# lets now slightly change the function and run the code again
def increase_age(user_age):
    user_age += 1


print(age)
print(id(age))
increase_age(age)
print(age)
print(id(age))
```

```python
# user_age += 1 quietly does the same thing as user_age = user_age + 1: new object, rebind the local label.
# Compare with a list:
ages = [20]


def increase_age(l):
    l += [1]


increase_age(ages)  # ages → [20, 1]  — list += mutates the shared object
print(ages)
#  The code looks the same, but the result is the opposite. The type decides it: mutable objects get changed in place, immutable ones get rebound.
```
