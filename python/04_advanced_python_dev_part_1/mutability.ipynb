# Mutability in Python

```python
# data once created if can be changed, it is considered mutable and if it can not be changed, it is considered immutable.

# for example, lets say if we create the below dict in python

friends = {"Jon": 1, "Ron": 2, "Amen": 3}

# here, data is the dictionary and not the variable i.e friends. When we create one such variable, we are simply creating a dict in the memory and giving it a name of friends.

# now, when we do

friends_dict = friends
# both these are pointing to the same dictionary that we had created above.

# we know that it is possible for us to take out the id of an object in python. this id is the address of the memory where that object is stored. If we print the ids of the above two, we should get the same ids.
print(id(friends))
print(id(friends_dict))
```

```python
# as we can see that the IDs of the above two objects is same.

# now lets say if we create another dict, with the same name and same information, let's see what id is given to it.

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))
```

```python
# while now we had the same variable name and this variable was pointing to the same type of data as well, their ids were different. This is because, python program runs in a sequential manner.

# when we create the first variable friends, basically we are creating a dict with the information of the friends and storing it in memory. Then python gives it a name of friends. Now, we must understand that there is a memory id 4459509952 and it has been allocated to a dict, or in other words it is not available or is occupied now.

# then we move ahead and run the same command again, where another dict even though with the same values, is created and stored in the memory. then a same name is also given to it. Now, however, when we print its id, it is different from the previous one. this is because these are two different objects, located at two different memory locations.

# now lets use the same dict again

friends = {"Jon": 1, "Ron": 2, "Amen": 3}
print(id(friends))

# now, we do the following
friends["Amen"] = 10
print(id(friends))
```

```python
# now, when we executed friends['Amen'] = 10, we did not make a new object. we simply modified the existing dict or in other words, we mutated the existing dict. hence, we can also say that a dictionary in python is a mutable data type or it can be modified after it is created.

# there are 4 data types which are not mutable in python, or data types, which once created can not be edited. any edit to these data types will return in creation of a new data altogether. these are -> int, float, strings and tuples. let us understand more using below examples:

int_1 = 1
# we have created int data type and stored it in memory, gave it the name of int_1. Now, lets check it's id
print(id(int_1))
```

```python
# now if we make any changes to the int_1, lets say, by adding 1 more to it
int_1 = int_1 + 1
# now, let us check its id.
print(id(int_1))
```

```python
# as we can see, when 1 became 2, since we had added 1 to it, it got stored at a new memory location, since, the original 1 could not be changed. same, as we mentioned is the case with strings, floats and tuples. So, when ever we make any changes to an existing int, str, float or a tuple, a new item of the same data type gets generated and is stored at a diff place in the memory. it happens, since these data types are immutable.

# now, since we are aware of what is mutability, let us understand why it is dangerous or what all do we need to take care of when working in python to prevent bugs related to mutability.

# in python, now since we know that variables are not buckets that contain data, but labels, which point to data or an object which is present in the memory.

# now, multiple variable names or labels can point to the same object in memory. If we change the data using one label, the data which other labels point to also changes. This happens with data which is mutable in nature, i.e other than int, str, float and tuples.

# let us now look at most common ways mutability causes bugs in python.

# 1. Aliasing
# suppose we have a list of numbers assigned to variable numb1. we then assign numb1 values to numb2. Now, both numb1 and numb2 are pointing to the same list, like below:
numb1 = [1, 2, 3]
numb2 = numb1

print(id(numb1))
print(id(numb2))
```

```python
# as we can see, both these variables point to the same object in memory. Now, if we make changes to the data using numb1, the same changes will reflect in numb1 also.

numb2.append(10)
```
